```python
import numpy as np
import pandas as pd

# Set seed for reproducibility
np.random.seed(42)

# Generate sales data (12 months × 4 categories)
sales_data = np.random.randint(1000, 5001, size=(12, 4))

sales_data
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

columns = ["Electronics", "Clothing", "Home & Kitchen", "Sports"]

df = pd.DataFrame(sales_data, columns=columns, index=months)

df
```

```python
print(df.head())
print("\nSummary Statistics:\n")
print(df.describe())
```

```python
category_totals = df.sum()
monthly_totals = df.sum(axis=1)

print("Category Totals:\n", category_totals)
print("\nMonthly Totals:\n", monthly_totals)
```

```python
growth = df.pct_change().mean()

print("Average Growth per Category:\n", growth)
```

```python
df["Total Sales"] = df.sum(axis=1)
df["Growth Rate"] = df["Total Sales"].pct_change()

df
```

```python
roll_number = 7  # CHANGE THIS

if roll_number % 2 == 0:
    df["Electronics"] *= 0.9
else:
    df["Clothing"] *= 0.85

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))

for col in columns:
    plt.plot(df.index, df[col], marker='o', label=col)

plt.title("Monthly Sales Trends")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df[columns])

plt.title("Sales Distribution by Category")
plt.show()
```

```python
arr = np.array([[1, -2, 3],
                [-4, 5, -6]])

# Absolute
abs_arr = np.abs(arr)

# Percentiles
percentiles = np.percentile(arr.flatten(), [25, 50, 75])

# Stats
mean = np.mean(arr)
median = np.median(arr)
std = np.std(arr)

print("Absolute:\n", abs_arr)
print("Percentiles:", percentiles)
print("Mean:", mean, "Median:", median, "Std Dev:", std)
```

```python
arr = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print("Floor:", np.floor(arr))
print("Ceil:", np.ceil(arr))
print("Truncate:", np.trunc(arr))
print("Round:", np.round(arr))
```
