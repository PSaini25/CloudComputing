```python
import random

rows = 20000
cols = 10000

with open("data.txt", "w") as f:
    for _ in range(rows):
        row = [str(random.randint(0, 100)) for _ in range(cols)]
        f.write(",".join(row) + "\n")
```

```python
def replace_with_first_row(matrix):
    first_row = matrix[0]
    for i in range(1, len(matrix)):
        matrix[i] = first_row[:]
    return matrix

# Example
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(replace_with_first_row(matrix))
```

```python
from collections import defaultdict

def group_elements(matrix):
    groups = defaultdict(list)

    for row in matrix:
        for val in row:
            groups[val].append(val)

    return list(groups.values())

matrix = [
    [1, 2, 2],
    [3, 1, 3]
]

print(group_elements(matrix))
```

```python
def add_matrices(A, B):
    return [[A[i][j] + B[i][j] for j in range(len(A[0]))] for i in range(len(A))]

def subtract_matrices(A, B):
    return [[A[i][j] - B[i][j] for j in range(len(A[0]))] for i in range(len(A))]

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

print("Addition:", add_matrices(A, B))
print("Subtraction:", subtract_matrices(A, B))
```

```python
def create_matrix(n):
    matrix = [[2 * (i + j) for j in range(n)] for i in range(n)]
    return matrix

n = 4
for row in create_matrix(n):
    print(row)
```

```python
def row_wise_sum(matrix):
    return [sum(row) for row in matrix]

matrix = [
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
]

print(row_wise_sum(matrix))
```
